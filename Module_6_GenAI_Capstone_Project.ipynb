```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

print("Gemini API connected successfully!")
```

```python
def study_assistant(question):
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=f"""
You are an AI Study Assistant.

Help the student understand the topic in simple and clear language.

Student's question:
{question}
"""
    )
    return response.text


question = input("Ask your study question: ")

print("\nAI Study Assistant Answer:\n")
print(study_assistant(question))
```

```python
def calculator(expression):
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return f"The answer is: {result}"
    except Exception:
        return "Sorry, I could not calculate that expression."

expression = input("Enter a calculation: ")

print(calculator(expression))
```

```python
def smart_study_assistant(question):
    question_lower = question.lower()

    # Decide whether to use calculator
    if any(op in question for op in ["+", "-", "*", "/"]):
        return calculator(question)

    # Otherwise use Gemini AI
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=f"""
You are an AI Study Assistant.
Explain the answer in simple and clear language.

Student question:
{question}
"""
    )
    return response.text


question = input("Ask anything: ")

print("\nAI Study Assistant:\n")
print(smart_study_assistant(question))
```

```python
!pip install -q pypdf
from google.colab import files
import pypdf

# Upload PDF
uploaded = files.upload()

# Get uploaded file name
pdf_name = list(uploaded.keys())[0]

# Read PDF
reader = pypdf.PdfReader(pdf_name)

text = ""

for page in reader.pages:
    extracted = page.extract_text()
    if extracted:
        text += extracted + "\n"

print("PDF uploaded successfully!")
print("Total characters:", len(text))
```

```python
# Ask questions about the PDF using Gemini

question = input("Ask a question about the PDF: ")

prompt = f"""
Answer the following question using ONLY the information from this PDF.

PDF Content:
{text}

Question: {question}
"""

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

print("\nAnswer:")
print(response.text)
```

```python
while True:
    question = input("\nAsk a question about the PDF (type 'exit' to stop): ")

    if question.lower() == "exit":
        print("Thank you for using the PDF AI Assistant!")
        break

    prompt = f"""
Answer the following question using ONLY the information from this PDF.

PDF Content:
{text}

Question: {question}
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    print("\nAnswer:")
    print(response.text)
```

```python
import gradio as gr

def ask_pdf(question):
    prompt = f"""
You are an AI Study Assistant.

Answer the question using ONLY the information from the uploaded PDF.

PDF Content:
{text}

Question:
{question}
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text


demo = gr.Interface(
    fn=ask_pdf,
    inputs=gr.Textbox(
        label="Ask a Question",
        placeholder="Ask something about your PDF..."
    ),
    outputs=gr.Textbox(
        label="AI Study Assistant Answer"
    ),
    title="📚 AI Study Assistant",
    description="Ask questions and get answers based on your uploaded PDF."
)

demo.launch(share=True)
```
